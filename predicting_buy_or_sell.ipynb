```python
import joblib
import pandas as pd

model = joblib.load("usduzs_buysell.pkl")
df = pd.read_csv('USD_UZS_data.csv')
df['date'] = pd.to_datetime(df['date'])
df = df.sort_values('date').reset_index(drop=True)

close_yesterday = df['close'].iloc[-1]
close = '12200.30'
change = (float(close) - close_yesterday) / close_yesterday * 100
close_lag1 = df['close'].iloc[-1]
close_lag7 = df['close'].iloc[-7]
close_ma7 = df['close'].iloc[-7:].mean()
change_lag1 = df['change'].iloc[-1]
dayofweek = 4  # Friday
month = 1  # January

my_date = pd.DataFrame([{
    'close_lag1': close_lag1,
    'close_lag7': close_lag7,
    'close_ma7': close_ma7,
    'change_lag1': change_lag1,
    'change': change,
    'dayofweek': dayofweek,
    'month': month
}])

predict_close = model.predict(my_date)

if predict_close[0] == 1:
    print("Predicted action: BUY")
else:
    print("Predicted action: SELL")
```
